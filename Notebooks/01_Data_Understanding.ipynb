## Step 2: Load the CSV Dataset Using Pandas

We use a relative path (`../Dataset/train.csv`) from the `Notebooks/` folder so the project remains portable. `pd.read_csv()` reads the CSV file into a Pandas DataFrame named `df`.

## Step 8 & Step 9: Identify Numerical vs. Categorical Features & Inspect Unique Values

Important domain insight:
- Some columns have `int64` or `float64` data types, yet represent **categories or binary flags** rather than continuous measurements:
  - `Driving_License` (`0`/`1`), `Previously_Insured` (`0`/`1`), `Response` (`0`/`1` target)
  - `Region_Code` and `Policy_Sales_Channel` (anonymized numeric codes representing categories)
- `id` is a unique row identifier, not a predictive feature.
- Truly continuous/discrete numerical features are `Age`, `Annual_Premium`, and `Vintage`.

```python
# Separate columns by Pandas dtype first
numeric_dtype_cols = df.select_dtypes(include=[np.number]).columns.tolist()
object_dtype_cols = df.select_dtypes(include=['object']).columns.tolist()

print('Columns with numeric dtypes:', numeric_dtype_cols)
print('Columns with object dtypes :', object_dtype_cols)

# Categorize columns by their real-world meaning
id_col = ['id']
target_col = ['Response']
continuous_numerical_cols = ['Age', 'Annual_Premium', 'Vintage']
binary_flag_cols = ['Driving_License', 'Previously_Insured']
categorical_code_cols = ['Region_Code', 'Policy_Sales_Channel']
categorical_text_cols = ['Gender', 'Vehicle_Age', 'Vehicle_Damage']

print('\n--- Feature Grouping by Domain Meaning ---')
print('Identifier              :', id_col)
print('Target Variable         :', target_col)
print('Continuous Numerical    :', continuous_numerical_cols)
print('Binary Flags (0/1)      :', binary_flag_cols)
print('Categorical Codes (num) :', categorical_code_cols)
print('Categorical Text (str)  :', categorical_text_cols)
```

```python
# Check unique value counts across all columns
print('--- Number of Unique Values per Column ---')
for col in df.columns:
    print(f'{col:22s}: {df[col].nunique():,}')

# Inspect exact unique values and value counts for low-cardinality categorical/binary columns
inspect_cols = categorical_text_cols + binary_flag_cols + target_col
for col in inspect_cols:
    print(f'\nValue counts for {col}:')
    vc = df[col].value_counts()
    pct = df[col].value_counts(normalize=True) * 100
    summary = pd.DataFrame({'Count': vc, 'Percentage (%)': pct.round(2)})
    display(summary)
```

## Step 10 & Step 11: Meaning of Each Feature & Target Variable

1. **`id`**: Unique customer identifier (1 to 381,109). Not useful for prediction.
2. **`Gender`**: Customer gender (`Male` or `Female`).
3. **`Age`**: Customer age in years.
4. **`Driving_License`**: `1` if the customer has a driving licence, `0` if not.
5. **`Region_Code`**: Numeric code identifying the customer's geographic region.
6. **`Previously_Insured`**: `1` if the customer already holds vehicle insurance, `0` otherwise.
7. **`Vehicle_Age`**: Age bracket of the customer's vehicle (`< 1 Year`, `1-2 Year`, `> 2 Years`).
8. **`Vehicle_Damage`**: Whether the customer's vehicle was previously damaged (`Yes` or `No`).
9. **`Annual_Premium`**: Yearly health insurance premium paid by the customer.
10. **`Policy_Sales_Channel`**: Anonymized code representing the sales/outreach channel (e.g., agent, online, phone).
11. **`Vintage`**: Number of days the customer has been associated with the insurance company.
12. **`Response` (Target Variable)**: `1` = Customer is interested in vehicle insurance, `0` = Customer is not interested.

## Step 12: Missing Values and Duplicate Records Check

We check `df.isnull().sum()` for missing values and `df.duplicated().sum()` for exact duplicate rows.

```python
# Check missing values per column
missing_counts = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_df = pd.DataFrame({'Missing_Count': missing_counts, 'Missing_Pct (%)': missing_pct})
print('--- Missing Values Summary ---')
display(missing_df)

# Check exact duplicate rows and duplicate customer IDs
print(f'Exact duplicate rows: {df.duplicated().sum()}')
print(f'Duplicate id values : {df["id"].duplicated().sum()}')
```
